```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

fontsize = 15
def plot_spectrum(ax, composition, x_values, y_values, xmin, xmax, xlabel, ylabel,
                  title, color_func, foldername,trajfolder, temp, timesteps, nsamples,
                  interval, std, freq_scale, offset_distance,minor_loc,ylim_factor):
    last_max = 0
    xspace = np.linspace(0, 500, 5000)
    freq_lim = 700
    nsteps = 10000
    for i, (x, y) in enumerate(zip(x_values, y_values)):
        
        color = color_func(x, y)
        
        #Plot experimental spectra
        data = pd.read_csv(f'{foldername}/{composition}/x{x:.3f}y{y:.3f}.csv', header=None)
        minmax_d = min(data[1]) - last_max
        xdata_expt = data[0].values
        ydata_expt = (data[1] - minmax_d + offset_distance).values
        last_max = max(ydata_expt)
        
        ax.fill_between(xdata_expt, ydata_expt, min(ydata_expt), alpha=0.4, color=color)
        
        if composition in ['MoWS2','MoWSe2']:
            ax.text(xmin + 0.03*(xmax-xmin), ydata_expt[0] + 1, f'x = {x}')
        else:
            ax.text(xmin + 0.03*(xmax-xmin), ydata_expt[0] + 1, f'y = {y}')
        
        # Calculate theoretical spectra
        
        
        nsamples = len(glob(f'{trajfolder}/x{x:.3f}y{y:.3f}T{temp:04.0f}_1fs_i{timesteps}*.npz'))
        freqs, avg_intens = calculate_average_spectra_new(trajfolder, x, y, temp, timesteps,nsteps,
                                                          nsamples, interval, xspace,freq_lim, std)
        
        intens_scale = (max(data[1]) - min(data[1])) / max(avg_intens)
        xdata_theo = xspace * freq_scale
        ydata_theo = avg_intens * intens_scale + min(ydata_expt)
        
        ax.plot(xdata_theo, ydata_theo, c=color)
    
    ax.set_xlim(xmin, xmax)
    ax.set_ylim(top = last_max + ylim_factor*(last_max - min(ydata_expt)))
    ax.set_yticks([])
    ax.xaxis.set_minor_locator(MultipleLocator(minor_loc))
    ax.text(0.02, 0.95, title, fontsize=14, fontweight='extra bold', transform=ax.transAxes)

std = 2
# Parameters for each composition
compositions = [
    {"name": "MoWS2", "x_values": [0.00, 0.20, 0.42, 0.53, 0.61, 0.77, 1.0], "y_values": [0] * 7, "xmin": 250, "xmax": 470, 
     "title":" Mo$_{{1-x}}$W$_{{x}}$S$_{{2}}$","std":std,"minor_loc":5,"ylim_factor":0.5},
    {"name": "MoWSe2", "x_values":  [0.00,0.21,0.39,0.53,1.0], "y_values": [1] * 7, "xmin": 225, "xmax": 260,
    "title":"Mo$_{{1-x}}$W$_{{x}}$Se$_{{2}}$","std":std,"minor_loc":5,"ylim_factor":0.5},
    {"name": "MoSSe", "x_values": [0] * 7, "y_values": [0.00, 0.28, 0.33, 0.47, 0.52, 0.67, 1.0], "xmin": 130, "xmax": 490,
    "title" :"MoS$_{{2-2y}}$Se$_{{2y}}$","std":std,"minor_loc":10,"ylim_factor":1.5},
    {"name": "WSSe", "x_values": [1] * 5, "y_values": [0.00, 0.30, 0.5, 0.7, 1.0], "xmin": 100, "xmax": 450,
     "title":"WS$_{{2-2y}}$Se$_{{2y}}$","std":std,"minor_loc":10,"ylim_factor":0.5}]
fontsize = 12
plt.rcParams['font.size'] = fontsize
fig, axs = plt.subplots(2, 2, figsize=(13, 12))
axs = axs.flatten()

foldername = '../quaternary_test/raman_test/raman_data_converged_new_version'
trajfolder = 'md_traj_pols'
for ax, comp in zip(axs, compositions):
    plot_spectrum(
        ax, comp["name"], comp["x_values"], comp["y_values"], comp["xmin"], comp["xmax"],
        r'Raman shift (cm$^{-1}$)', 'Intensity (a.u.)',fr'{comp["title"]}',get_alloy_color,
        foldername,trajfolder,temp=300, timesteps=10000, nsamples=15, interval=10, std=comp["std"],
        freq_scale=1.035, offset_distance=2,minor_loc=comp["minor_loc"],ylim_factor=comp["ylim_factor"])
    
fig.text(0.5, 0.001, r"Raman shift (cm$^{-1}$)", ha='center',fontsize=fontsize+2)
fig.text(0.0001, 0.5, 'Intensity (a.u.)', va='center', rotation='vertical',fontsize=fontsize+2)

plt.tight_layout()
plt.show()
```

```python
temp = 300
timesteps = 10000
nsteps = 10000
nsamples = 50
trajfolder = 'md_traj_pols/'

freq_lim = (100,700)
std = 3
interval = 1
freq_scale_full = 1.045
freq_scale_first = 1.035

fontsize = 16
plt.rcParams['font.size'] = fontsize
fig,axs = plt.subplots(5,3,figsize=(12,12))


comps = np.array([[[0,1],[0.53,1],[1,1]],
         [[0,0.67],[0.25,0.75],[1,0.7]],
         [[0,0.52],[0.5,0.5],[1,0.5]],
         [[0,0.28],[0.75,0.25],[1,0.3]],
        [[0,0],[0.53,0],[1,0]]])


folder_expt = '../quaternary_test/raman_test/raman_data_converged_new_version'
folder_first = 'raman_data_dielectrics_w10_1/'

for i in range(axs.shape[0]):
    for j in range(axs.shape[1]):
        xspace = np.linspace(0, 500, 5000)
        x,y = comps[i][j]
        color = get_alloy_color(x,y)
        
        if j==0:
            composition = 'MoSSe'
            xmin,xmax =  130,490
        elif j == axs.shape[1]-1:
            composition = 'WSSe'
            xmin,xmax = 100,450
        elif i == 0:
            composition ='MoWSe2'
            xmin, xmax = 100,450
        elif i == axs.shape[0]-1:
            composition = 'MoWS2'
            xmin, xmax = 100,450
        else:
            composition = 'MoWSSe'
            xmin,xmax = 150,450
        
        if not (x == 0.25 or x == 0.75):
            
            data = pd.read_csv(f'{folder_expt}/{composition}/x{x:.3f}y{y:.3f}.csv', header=None)
            xdata_expt = data[0].values
            ydata_expt = (data[1]).values
            
            axs[i][j].fill_between(xdata_expt, ydata_expt, min(ydata_expt), alpha=0.4, color=color)
        
        #Full Raman
        freqs, avg_intens = calculate_average_spectra(trajfolder, x, y, temp, timesteps,nsteps,
                                                          nsamples, interval, xspace,freq_lim, std)
        freqs*= freq_scale_full
        if not (x == 0.25 or x == 0.75):
            intens_scale = (max(data[1]) - min(data[1])) / max(avg_intens)
            avg_intens = avg_intens * intens_scale + min(ydata_expt)
            
        axs[i][j].plot(freqs,avg_intens,color=color)
        
        
        
        #First order Raman
        nconfig = 1 if (y==0.0 or y==1.0) else 15
        avg_intens_static = np.zeros(len(xspace))
        for config in range(nconfig):
            npzfile = np.load(f'{folder_first}/{composition}_x{x:.3f}y{y:.3f}_config_{config}.npz')
            intens_static,freqs_static = npzfile['intensities'],npzfile['freqs']
            avg_intens_static = avg_intens_static + get_spectra_curve(xspace,intens_static,freqs_static,std)[1]
        
        avg_intens_static = avg_intens_static[1:]
        
        avg_intens_static /= nconfig
        c = 299792458
        xspace = xspace[1:]
        static_prefactor = get_prefactor(xspace*(c * 1e2),temp)
        avg_intens_static *= static_prefactor
        

        
        freqs_static = freq_scale_first*xspace
        
        #Get prefactor for static intensity
        
        
        
        if not (x == 0.25 or x == 0.75):
            intens_scale = (max(data[1]) - min(data[1])) / max(avg_intens_static)
            avg_intens_static = avg_intens_static * intens_scale + min(ydata_expt)
        else:
            intens_scale = (max(avg_intens) - min(avg_intens)) / max(avg_intens_static)
            avg_intens_static = avg_intens_static * intens_scale + min(avg_intens)
        
        axs[i][j].plot(freqs_static,np.real(avg_intens_static),linestyle='--',lw=1,color=color)
        
        if i!=4:
            axs[i][j].tick_params(labelbottom=False)
        

        axs[i][j].set_xlim(xmin,xmax)
        

        label = ""
        if x != 1:
            label += rf'Mo$_{{{1-x:.3g}}}$' if x!=0 else 'Mo'
        if x != 0:
            label += rf'W$_{{{x:.3g}}}$' if x!=1 else 'W'
        if y != 1:
            if y==0:
                label += r'S$_{2}$'
            elif y==0.5:
                label += 'S'
            else:
                label += rf'S$_{{{2-2*y:.3g}}}$' 
        if y != 0:
            if y==1:
                label += r'Se$_{2}$'
            elif y==0.5:
                label += 'Se'
            else:
                label += rf'Se$_{{{2*y:.3g}}}$'
            
            
        axs[i][j].text(0.05, 0.95, label, transform=axs[i][j].transAxes,
                       ha='left', va='top')
        #print(x,y)

        axs[i][j].set_yticks([])
        
        axs[i][j].set_ylim(top = max(avg_intens) + 0.3*(max(avg_intens)-min(avg_intens)))
        axs[i][j].xaxis.set_minor_locator(MultipleLocator(10))

plt.subplots_adjust(wspace=0.05, hspace=0.05)
    

fig.text(0.5, 0.075, r"Raman shift (cm$^{-1}$)", ha='center',fontsize=fontsize+2)
fig.text(0.09, 0.5, 'Intensity (a.u.)', va='center', rotation='vertical',fontsize=fontsize+2)   

plt.savefig('alloy_raman_combined.png',facecolor='white', transparent=False, dpi=300, bbox_inches='tight')

plt.show()
```

```python
temp = 300
timesteps = 10000
nsteps = 10000
nsamples = 50
trajfolder = 'md_traj_pols/'

freq_lim = (50,700)
std_full = 3
std_first = 3
interval = 1
freq_scale_full = 1.045
freq_scale_first = 1.035

fontsize = 9.5

plt.rcParams['font.size'] = fontsize
fig, axs = plt.subplots(3, 5, figsize=(9, 4.5))  # transposed layout

comps = np.array([[[0,1],[0.53,1],[1,1]],
                  [[0,0.67],[0.25,0.75],[1,0.7]],
                  [[0,0.52],[0.5,0.5],[1,0.5]],
                  [[0,0.28],[0.75,0.25],[1,0.3]],
                  [[0,0],[0.53,0],[1,0]]])

folder_expt = '../quaternary_test/raman_test/raman_data_converged_new_version'
folder_first = 'raman_data_dielectrics_w10_1/'

for i in range(5):  # original rows
    for j in range(3):  # original columns
        #ax = axs[2-j][4-i]  # transposed index
        ax = axs[j][4-i]
        xspace = np.linspace(0, 500, 5000)
        x, y = comps[i][j]
        color = get_alloy_color(x, y)

        if j == 0:
            composition = 'MoSSe'
            xmin, xmax = 100, 480
        elif j == 2:
            composition = 'WSSe'
            xmin, xmax = 100, 480
        elif i == 0:
            composition = 'MoWSe2'
            xmin, xmax = 100, 480
        elif i == 4:
            composition = 'MoWS2'
            xmin, xmax = 100, 480
        else:
            composition = 'MoWSSe'
            xmin, xmax = 100, 480

        if not (x == 0.25 or x == 0.75):
            data = pd.read_csv(f'{folder_expt}/{composition}/x{x:.3f}y{y:.3f}.csv', header=None)
            xdata_expt = data[0].values
            ydata_expt = data[1].values
            ax.fill_between(xdata_expt, ydata_expt, min(ydata_expt), alpha=0.4, color=color)

        # Full Raman spectra
        freqs, avg_intens = calculate_average_spectra(
            trajfolder, x, y, temp, timesteps, nsteps, nsamples,
            interval, xspace, freq_lim, std_full
        )
        freqs *= freq_scale_full
        if not (x == 0.25 or x == 0.75):
            intens_scale = (max(data[1]) - min(data[1])) / max(avg_intens)
            avg_intens = avg_intens * intens_scale + min(ydata_expt)

        ax.plot(freqs, avg_intens, color=color)

        # First-order Raman
        nconfig = 1 if (y == 0.0 or y == 1.0) else 15
        avg_intens_static = np.zeros(len(xspace))
        for config in range(nconfig):
            npzfile = np.load(f'{folder_first}/{composition}_x{x:.3f}y{y:.3f}_config_{config}.npz')
            intens_static, freqs_static = npzfile['intensities'], npzfile['freqs']
            avg_intens_static = avg_intens_static + get_spectra_curve(xspace, intens_static, freqs_static, std_first)[1]

        
        avg_intens_static = avg_intens_static[1:]
        avg_intens_static /= nconfig
        xspace = xspace[1:]
        c = 299792458
        static_prefactor = get_prefactor(xspace * (c * 1e2), temp)
        avg_intens_static *= static_prefactor
        freqs_static = freq_scale_first * xspace

        if not (x == 0.25 or x == 0.75):
            intens_scale = (max(data[1]) - min(data[1])) / max(avg_intens_static)
            avg_intens_static = avg_intens_static * intens_scale + min(ydata_expt)
        else:
            intens_scale = (max(avg_intens) - min(avg_intens)) / max(avg_intens_static)
            avg_intens_static = avg_intens_static * intens_scale + min(avg_intens)

        ax.plot(freqs_static, np.real(avg_intens_static), linestyle='--', lw=1.3, color=color)

        if j != 2:
            ax.tick_params(labelbottom=False)

        ax.set_xlim(xmin, xmax)
        ax.set_yticks([])

        label = ""
        if x != 1:
            label += rf'Mo$_{{{1-x:.3g}}}$' if x != 0 else 'Mo'
        if x != 0:
            label += rf'W$_{{{x:.3g}}}$' if x != 1 else 'W'
        if y != 1:
            if y == 0:
                label += r'S$_{2}$'
            elif y == 0.5:
                label += 'S'
            else:
                label += rf'S$_{{{2-2*y:.3g}}}$'
        if y != 0:
            if y == 1:
                label += r'Se$_{2}$'
            elif y == 0.5:
                label += 'Se'
            else:
                label += rf'Se$_{{{2*y:.3g}}}$'

        ax.text(0.02, 0.95, label, transform=ax.transAxes,
                ha='left', va='top')

        ax.set_ylim(top=max(avg_intens) + 0.3 * (max(avg_intens) - min(avg_intens)))
        ax.xaxis.set_minor_locator(MultipleLocator(10))

plt.subplots_adjust(wspace=0.05, hspace=0.05)
fig.text(0.5, 0.04, r"Raman shift (cm$^{-1}$)", ha='center', fontsize=fontsize+2)
fig.text(0.1, 0.5, 'Intensity (a.u.)', va='center', rotation='vertical', fontsize=fontsize+2)

plt.savefig('alloy_raman_combined_transposed.png', facecolor='white', transparent=False, dpi=300, bbox_inches='tight')
plt.show()
```
