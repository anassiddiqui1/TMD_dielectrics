```python
from utils import *
#from utils_old import *
%matplotlib inline
import matplotlib.colors as mcolors
import matplotlib.ticker as ticker
```

```python
fontsize = 15
plt.rcParams['font.size'] = fontsize
fig, axes = plt.subplots(2, 2, figsize=(10, 10), sharex='col')
temps = [100,300,500,700]
comps = ((0,0),(0,1),(1,0),(1,1))
comps_labels = {(0,0):'MoS$_2$',(1,0):'WS$_2$',(0,1):'MoSe$_2$',(1,1):'WSe$_2$'}

foldername = 'md_traj_pols'
timesteps = 10000
nruns = 15
interval = 1
std = 3
nsteps = 10000
freq_lim = (10,1000)
xspace = np.linspace(0,1000,10000)

# Choose a colormap that reflects temperature (inferno, plasma, or hot)
base_cmap = plt.cm.plasma  # Other options: plt.cm.hot, plt.cm.plasma
cmap = mcolors.LinearSegmentedColormap.from_list(
    "custom_inferno", base_cmap(np.linspace(0, 0.7, 256))  # Clip to avoid the lightest colors
)
norm = plt.Normalize(vmin=min(temps), vmax=max(temps))  # Normalize temperatures
axes = axes.flatten()  # Flatten the axes array for easier iteration


for idx,(x,y) in enumerate(comps):
    ax = axes[idx]
    
    offset = 0
    last_max = 0
    
    if y==1:
        xmin,xmax = 180,350     
    else:
        xmin,xmax = 300,450

    for temp in temps:
        freqs,avg_intens = calculate_average_spectra(foldername,x,y,temp,timesteps,nsteps,nruns,
                                                    interval,xspace,freq_lim,std)
        color = cmap(norm(temp))
        xdata_theo = freqs*1.035
        ydata_theo = offset+last_max+avg_intens/temp
        
        ax.plot(xdata_theo,ydata_theo,label = f'{temp} K',color=color,alpha=0.5)
        ax.text(0.01, 0.98, f'{comps_labels[(x,y)]}', transform=ax.transAxes,
            fontsize=20, verticalalignment='top', horizontalalignment='left')
        
        last_max = max(ydata_theo)
        #ax.text(xmin + 0.01*(xmax-xmin), ydata_theo[0] + 0.2*(max(ydata_theo)-min(ydata_theo)), f'{temp} K')
        
        
        if temp == temps[0]:
            offset = 0.5*max(ydata_theo)
        
        if idx == 0:
            handles, labels = ax.get_legend_handles_labels() 
        ax.set_yticks([])
    
    ax.set_ylim(top = last_max + 0.3*(last_max-min(ydata_theo)))
    ax.set_xlim(xmin,xmax)
    ax.xaxis.set_major_locator(MultipleLocator(50))
    ax.xaxis.set_minor_locator(MultipleLocator(5))

fig.legend(handles, labels, loc='upper center', fontsize=fontsize,
           ncol=len(labels), frameon=False, bbox_to_anchor=(0.5, 0.93),
           prop={'weight': 'bold'},columnspacing=2)

fig.text(0.5, 0.06, r"Raman shift (cm$^{-1}$)", ha='center',fontsize=fontsize+2)
fig.text(0.08 ,0.5, "Intensity (a.u.)", va='center', rotation='vertical',fontsize=fontsize+2)

plt.subplots_adjust(wspace=0.15, hspace=0.1)
plt.show()
```

# Plot experimental comparison for pure MoS$_2$, MoSe$_2$ and WSe$_2$ at multiple temperatures

```python
fontsize = 16

temps = [175,225,300,375,475,575]
comps = ((0,0),(0,1),(1,1))
comps_labels = {(0,0):'MoS$_2$',(1,0):'WS$_2$',(0,1):'MoSe$_2$',(1,1):'WSe$_2$'}
freq_scale = 1.045


foldername = 'md_traj_pols' #Folder storing all .traj files
timesteps = 10000 ## Number of total timesteps in each trajectory
nruns = 50 ## total number of trajectories to use
interval = 1 ## dt
std = 3  ##broadening parameter 
nsteps = 10000 ## Number of timesteps to use in each trajectory
freq_lim = (10,1000)
xspace = np.linspace(0,1000,10000)
npoly = 1

plt.rcParams['font.size'] = fontsize
fig, axes = plt.subplots(2, 2, figsize=(10, 10))#, sharex='col')
axes = axes.flatten()  # Flatten the axes array for easier iteration
plt.subplots_adjust(wspace=0.3, hspace=0.22)
pos = axes[-1].get_position()
fig.delaxes(axes[-1])
height = pos.height
gap = 0.01  # vertical gap between broken axes
ax_top = fig.add_axes([pos.x0, pos.y0 + height / 2 + gap / 2, pos.width, height / 2 - gap / 2])
ax_top.set_ylim(375, 430)
ax_top.set_xlim(150,800)
ax_top.spines['bottom'].set_visible(False)
ax_top.tick_params(axis='x', which='both', bottom=False, top=False, labelbottom=False)

# Bottom part (lower y range)
ax_bot = fig.add_axes([pos.x0, pos.y0, pos.width, height / 2 - gap / 2])
ax_bot.set_ylim(230, 260)
ax_bot.set_xlim(150,800)
ax_bot.spines['top'].set_visible(False)
d = .01
kwargs = dict(transform=ax_top.transAxes, color='k', clip_on=False)
ax_top.plot((-d, +d), (-d, +d), **kwargs)
ax_top.plot((1 - d, 1 + d), (-d, +d), **kwargs)
kwargs.update(transform=ax_bot.transAxes)
ax_bot.plot((-d, +d), (1 - d, 1 + d), **kwargs)
ax_bot.plot((1 - d, 1 + d), (1 - d, 1 + d), **kwargs)

# Choose a colormap that reflects temperature (inferno, plasma, or hot)
base_cmap = plt.cm.plasma  # Other options: plt.cm.hot, plt.cm.plasma
cmap = mcolors.LinearSegmentedColormap.from_list(
    "custom_inferno", base_cmap(np.linspace(0, 0.7, 256))  # Clip to avoid the lightest colors
)
norm = plt.Normalize(vmin=min(temps), vmax=max(temps))  # Normalize temperatures

#Define xlimits for each suplot
x_lims = {(0,0):(360,420),(0,1):(200,300),(1,1):(200,300)}

for idx,(x,y) in enumerate(comps):
    ax = axes[idx]
    
    offset = 10
    last_max = 0
    xmin,xmax = x_lims[comps[idx]]

    #Arrays to track peaks
    A1_theory = []
    A1_expt = []
    E_theory = []
    E_expt = []
    
    #Arrays for guidelines
    A1_guide = []
    E_guide = []
    
    if idx!=0:
        A1_range_min,A1_range_max = 230,260
        E_range_min,E_range_max = 280,300
    else:
        A1_range_min,A1_range_max = 390,420
        E_range_min,E_range_max = 370,390

    for temp in temps:
        color = cmap(norm(temp))
        
        #Experimental spectra
        data = pd.read_csv(f'expt_temp/x{x:.3f}y{y:.3f}T{temp:04.0f}.csv',header=None)
        minmax_d = min(data[1])-last_max
        xdata_expt = data[0].values
        ydata_expt = (data[1]-minmax_d+offset).values
        last_max = max(ydata_expt)
        
        ax.fill_between(xdata_expt,ydata_expt,min(ydata_expt),alpha=0.4,color=color)
        ax.text(xmin + 0.01*(xmax-xmin), 
                ydata_expt[0] + 0.2*(max(ydata_expt)-min(ydata_expt)),
                f'{temp} K')
        
        #Calculated spectra
        freqs,avg_intens = calculate_average_spectra(foldername,x,y,temp,timesteps,nsteps,nruns,
                                                    interval,xspace,freq_lim,std)
        
        #Scaling intensities such that max of calculated is equal to max of experimental
        intens_scale = (max(data[1])-min(data[1]))/max(avg_intens)
        #intens_scale = (max(ydata_expt))/max(avg_intens)
        xdata_theo = xspace*freq_scale
        ydata_theo = avg_intens*intens_scale+min(ydata_expt)
        ax.plot(xdata_theo,ydata_theo,label = f'{temp} K',color=color)
        
        #print(intens_scale,max(ydata_theo)==max(ydata_expt))
        
        #Get A1' peak_positions
        A1_x,A1_y,A1_idx = get_max_peak(xdata_theo,ydata_theo,A1_range_min,A1_range_max,return_index=True)
        
        A1_theory.append(A1_x)
        A1_guide.append(A1_y)
        if temp == temps[-1]:
            ax.text(A1_x+2,A1_y+5,r"A$_{1}'$",ha='left',va='center')
        
        A1_x,A1_y = get_max_peak(xdata_expt,ydata_expt,A1_range_min,A1_range_max)
        A1_expt.append(A1_x)
        
        if comps[idx]==(0,0):
            E_x,E_y = get_max_peak(xdata_theo,ydata_theo,E_range_min,E_range_max)
            E_theory.append(E_x)
            E_guide.append(E_y)
            
            if temp == temps[-1]:
                ax.text(E_x+2,E_y+2,r"E$'$",ha='left',va='center')
            
            E_x,E_y = get_max_peak(xdata_expt,ydata_expt,E_range_min,E_range_max)
            E_expt.append(E_x)
            
        if comps[idx]==(0,1):
            E_x,E_y = get_max_peak(xdata_theo,ydata_theo,E_range_min,E_range_max)
            E_theory.append(E_x)
            E_guide.append(E_y)
            
            E_range_max = E_x+5
            
            if temp == temps[-1]:
                ax.text(E_x+2,E_y+2,r"E$'$",ha='left',va='bottom')

    poly = np.polyfit(A1_guide,A1_theory,npoly)
    ax.plot(np.polyval(poly,A1_guide),A1_guide,linestyle='--',c=get_alloy_color(x,y))

    if comps[idx]!=(1,1):
        poly = np.polyfit(E_guide,E_theory,npoly)
        ax.plot(np.polyval(poly,E_guide),E_guide,linestyle='--',c='r')

    ax.set_yticks([])
    ax.set_ylabel("Intensity (a.u.)")
    ax.set_xlabel(r"Raman shift (cm$^{-1}$)")
    
    ax.set_ylim(top = last_max + 2*(last_max-min(ydata_theo)))
    ax.set_xlim(xmin,xmax)
    ax.xaxis.set_minor_locator(MultipleLocator(5))
    ax.text(0.01, 0.98, f"({chr(ord('a')+idx)}) {comps_labels[(x,y)]}", transform=ax.transAxes,
    fontsize=fontsize, verticalalignment='top', horizontalalignment='left',fontweight='bold')
    
    if idx==0:
        ax = ax_top
        poly = np.polyfit(temps,E_theory,npoly)
        E_theory_plot = ax.plot(temps,np.polyval(poly,temps),marker='^',color='r')
        
        poly = np.polyfit(temps,E_expt,npoly)
        E_expt_plot = ax.plot(temps,np.polyval(poly,temps),marker='s',color='r',
                              linestyle='--',markerfacecolor='none')

        ax.text(temps[-1]+12,E_expt[-1],rf"E' {comps_labels[(x,y)]}")
    else:
        ax = ax_bot
        
    poly = np.polyfit(temps,A1_theory,npoly)
    A1_theory_plot = ax.plot(temps,np.polyval(poly,temps),marker='^',color=get_alloy_color(x,y))

    poly = np.polyfit(temps,A1_expt,npoly)
    A1_expt_plot = ax.plot(temps,np.polyval(poly,temps),marker='s',color=get_alloy_color(x,y),
                          linestyle='--',markerfacecolor='none')

    ax.text(temps[-1]+12,A1_expt[-1],rf"A$'_{1}$ {comps_labels[(x,y)]}")

    
experiment_line = mlines.Line2D([], [], color='black', marker='s',markerfacecolor='None',
                                linestyle='--',label='Expt.')
calculated_line = mlines.Line2D([], [], color='black', marker='s',label='Theory')
ax_top.legend(handles=[experiment_line,calculated_line],loc='upper right',
              frameon=False,ncols=2,columnspacing=0.5)
    
ax_top.text(0.01, 0.92, f"(d)", transform=ax_top.transAxes,
    fontsize=fontsize, verticalalignment='top', horizontalalignment='left',
           fontweight='bold')
        
ax_top.set_ylabel(r"Raman shift (cm$^{-1}$)")   
ax_top.yaxis.set_label_coords(-0.15, -0.1)
ax_bot.set_xlabel("Temperature (K)")

plt.savefig('raman_temps.png',facecolor='white', 
            transparent=False,dpi=300,bbox_inches='tight',pad_inches=0.02)

plt.show()
```
