# Convergence with respect to timestep

```python
fontsize = 15
plt.rcParams['font.size'] = fontsize
fig,axs = plt.subplots(figsize=(7,5))
foldername = 'monolayer_md_data/'
xspace = np.linspace(0,1200,10000)
x = 0.5
y = 0.5
temp = 300
timesteps = 10000
std = 3
interval = 1
intervals = [1,2,5,10,20,40] # Different timesteps
freq_lim = (50,800)

nruns = 50 
nsteps = 10000

for interval in intervals: 
    freqs,intens = calculate_average_spectra(foldername,x,y,temp,timesteps,
                                             nsteps,nruns,interval,xspace,
                                             freq_lim,std)
    intens /= max(intens)
    if interval == 1:
        ref_intens = intens
    axs.plot(freqs,np.abs(intens-ref_intens)/max_intens,label =rf'$dt$ = {interval} fs')
    
plt.legend(frameon=False)

# plt.title('Spectra vs Timestep')
axs.xaxis.set_minor_locator(MultipleLocator(10))
#axs.set_yticks([])
axs.set_xlim(freq_lim[0],freq_lim[1])
axs.set_xlabel(r"Raman shift (cm$^{-1}$)")
axs.set_ylabel(r"Scaled Intensity Error")

plt.savefig('spectra_timestep_convergence.png',dpi=300,facecolor='white',transparent=False,
            bbox_inches='tight',pad_inches=0.02)

plt.show()
```

# Convergence with respect to number of trajectories

```python
fontsize = 15
plt.rcParams['font.size'] = fontsize
fig,axs = plt.subplots(figsize=(7,5))
foldername = 'monolayer_md_data/'
xspace = np.linspace(0,1200,10000)
x = 0.5
y = 0.5
temp = 300

std = 3
interval = 1
freq_lim = (50,800)
timesteps = 10000
nsteps = 10000
nruns = [10,20,30,40,50,100][::-1]

for nrun in nruns: 
    freqs,intens = calculate_average_spectra(foldername,x,y,temp,timesteps,
                                             nsteps,nrun,interval,xspace,
                                             freq_lim,std)
    intens /= max(intens)
    if nrun == 100:
        ref_intens = intens
        max_intens = max(intens)

    axs.plot(freqs,np.abs(intens-ref_intens)/max_intens,label =rf'$N_{{traj}}$ = {nrun}')

axs.legend(frameon = False)
axs.set_xlim(freq_lim[0],freq_lim[1])
axs.set_xlabel(r"Raman shift (cm$^{-1}$)")
axs.set_ylabel(r"Scaled Intensity Error")
#axs.set_ylabel(r"Intensity (a.u.)")
#axs.set_yticks([])
axs.xaxis.set_minor_locator(MultipleLocator(10))

plt.savefig('spectra_nruns_convergence.png',dpi=300,facecolor='white',transparent=False,
            bbox_inches='tight',pad_inches=0.02)

plt.show()
```
